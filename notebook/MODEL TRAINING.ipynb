```python
X = df.drop('math_score',axis=1)
y = df['math_score']
print(X.shape)
print(y.shape)
```

```python
## spljt the data into train test split
```

```python
from sklearn.model_selection import train_test_split

X_train, y_train, X_test, y_test =train_test_split(X,y, test_size=0.20,random_state=42)

print(X_train.shape, y_train.shape)
```

```python
## applying tranformation which is 3 types
```

```python
num_feature = df.select_dtypes(include='number').columns
cat_feature = df.select_dtypes(exclude='number').columns
print(f"we have numerical feature:{num_feature}")
print(f"we have categorical feature:{cat_feature} ")
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
num_pipeline = Pipeline(
    steps=[
        ("num",SimpleImputer(strategy='median')),
        ("scaler",StandardScaler())
    ]
)

cat_pipeline = Pipeline(
    steps=[
        ("cat",SimpleImputer(strategy="most_frequent")),
        ("one_hot_encoder",OneHotEncoder())
    ]
)

preprocessor = ColumnTransformer(
    [
        ("num_pipe",num_feature,num_pipeline),
        ("cat_pipe",cat_feature,cat_pipeline)
    ]
)
```

```python
preprocessor
```

```python
X_train_transformer=preprocessor.fit_transform(X_train)
X_test_transformer =preprocessor.transform(X_test)
```
